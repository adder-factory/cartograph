# Analysis
Loads data and computes totals.


```python
import json
from collections import Counter
from helpers import normalize
```

```python
class Loader:
    def __init__(self, path):
        self.path = path

    def load(self):
        with open(self.path) as fh:
            return json.load(fh)
```

```python
def total(rows):
    return sum(normalize(r) for r in rows)

loader = Loader('data.json')
rows = loader.load()
print(total(rows))
```

```python
counts = Counter(r['kind'] for r in rows)
counts.most_common(3)
```
